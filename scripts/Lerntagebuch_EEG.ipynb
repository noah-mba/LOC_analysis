## Lerntagebuch

**Independent Component Analysis (ICA)** is a computational method for separating a multivariate signal into its original, statistically independent subcomponents.
ICA is used in EEG, because the electrodes record both brain activity and noise (like eye blinks or heartbeats). ICA separates the combined channel data back into its distinct, underlying source components. This allows us to clean the data without losing it. E.g., ICA should isolate blinks into a single component and we can drop this artefact to erconstruct the clean EEG signal.
