**Formatierte Strings (f-Strings)** erlauben Text und Code-Outputs in einer Print-Funktion zu kombinieren.

```python
print(f"Ich hatte 5 Äpfel. Dann habe ich 2 gegessen. Jetzt habe ich noch {5-2} Äpfel.")
print(f"Das Produkt aus 7 und 9 ergibt {7*9}.")
```

**Unpacking** ist ein Python-spezifisches Konzept und bedeutet, dass man mehrere Werte aus einer Datensammlung (wie einem Tupel oder einer Liste) in einem einzigen Schritt herausnehmen und direkt an separate Variablen zuweisen kann.

```python
# Eine Funktion gibt zwei Werte als Tupel zurück
mein_tupel = (10, 20)

# Manuelles Zuweisen über den Index (0 und 1)
a = mein_tupel[0]  # a wird 10
b = mein_tupel[1]  # b wird 20

# Beide Variablen werden in einer einzigen Zeile "entpackt"
c, d = (10, 20)

if a == c and b == d:
    print('Unpacking hat funktioniert!')
```

**Ein Objekt** ist das grundlegende "Ding" in Python als **objekt-orientierter Programmiersprache**. Tatsächlich ist in Python absolut alles ein Objekt; es ist schlichtweg ein Datenpaket, das seine eigenen Eigenschaften (Attribute) und Fähigkeiten (Methoden) mitbringt. Mit der *type()-Funktion* lässt sich herausfinden, um weche Art von Objekt es sich handelt.

```python
mein_tupel = (10, 20, 30)
print(f"{type(mein_tupel)} - Ein Tupel speichert mehrere, fest verbundene und unveränderliche Werte")

print(f"{type(3)} - Ein Integer ist eine ganzzahlige Zahl (ohne Komma)")

print(f"{type(3.14)} - Ein Float ist eine Gleitkommazahl (Zahl mit Dezimalstellen, wichtig: Punkt statt Komma!)")

text_beispiel = "Hallo EEG"
print(f"{type(text_beispiel)} - Ein String ist eine Zeichenkette (also reiner Text)")

liste_beispiel = ['Fz', 'Cz', 'Pz']
print(f"{type(liste_beispiel)} - Eine Liste speichert mehrere Werte, die im Nachhinein beliebig verändert, gelöscht oder erweitert werden können")

dict_beispiel = {'sfreq': 250, 'channels': 64}
print(f"{type(dict_beispiel)} - Ein Dictionary (Wörterbuch) speichert Daten als Schlüssel-Wert-Paare")

print(f"{type(True)} - Ein Boolean ist ein logischer Wahrheitswert (kennt nur True oder False)")
```

**Das magische Trio der objekt-orientierten Programmierung in Python**
1. **Klasse (Class)**: Der Bauplan für ein Objekt. Die Klasse legt als abstrakte Vorlage fest, welche Eigenschaften und Fähigkeiten zukünftige Objekte haben werden, die nach diesem Bauplan erstellt werden. Klassen werden meistens groß geschrieben am Anfang.

2. **Attribut (Attribute)**: Die Eigenschaften (Zustände oder Metadaten), die in einem konkreten Objekt gespeichert sind. Wichtig: Da hier keine Aktion ausgeführt wird, ruft man sie ohne Klammern am Ende auf!

3. **Methode (Method)**: Die Fähigkeiten (eingebaute Funktionen), die ein Objekt besitzt. Es sind Aktionen, die das Objekt mit sich selbst durchführen kann. Wichtig: Da hier etwas getan wird, ruft man sie immer mit Klammern () am Ende auf!

```python
# ==========================================
# 1. Das Trio in Basis-Python (Wir bauen eine eigene Klasse)
# ==========================================

# KLASSE: Wir definieren den Bauplan für einen "Probanden"
class Proband:
    # Hier legen wir fest, was beim Erstellen des Objekts passiert
    def __init__(self, id_nummer, alter):
        # ATTRIBUTE: Eigenschaften, die wir im Objekt speichern
        self.id = id_nummer
        self.alter = alter
    
    # METHODE: Eine Aktion, die unser Proband ausführen kann
    def daten_liefern(self):
        return f"Proband {self.id} liefert jetzt fleißig Daten."

# Wir erstellen ein konkretes Objekt nach unserem Bauplan
p1 = Proband(id_nummer="VP_001", alter=25)

print(f"Das Objekt entspricht folgender Klasse: {type(p1)}")

# Wir rufen ein ATTRIBUT ab (Eigenschaft -> keine Klammern!)
print(f"Eigenschaft abgerufen: Der Proband ist {p1.alter} Jahre alt.")

# Wir rufen eine METHODE auf (Aktion -> mit Klammern!)
print(f"Aktion ausgeführt: {p1.daten_liefern()}")
```

```python
# ==========================================
# 2. Wir untersuchen die Attribute und Methode einer Klasse
# ==========================================
alle_faehigkeiten = dir(Proband)
print(alle_faehigkeiten)
## Die Methoden mit doppelten Unterstrichen nennt man "Dunder"-Methoden (Double Underscore), die ich erstmal ignorieren darf.

# Wenn du nicht nur die Namen wissen willst, sondern auch, wie sie funktionieren:
help(Proband)

# Alternativ einfach den Namen des Objekts mit Punkt tippen und dann aus dem Drop-Down Menü auswählen
```

**Indexing** bezeichnet das gezielte Abrufen von einzelnen Werten in Python und beginnt stets bei 0. Mit **Slicing** meinen wir das Abrufen von mehreren Werten gleichzeitig. 
Die grundsätzliche Syntax für Listen und Tuples ist dabei die gleiche, während Dictionaries nur über Keys ansprechbar sind.

```python
# 1. LISTEN und TUPLES
# -----------------------------------------------
# Ersten Wert abrufen
print(mein_tupel[0])
print(liste_beispiel[0])

# Letzten Wert abrufen
print(mein_tupel[-1])
print(liste_beispiel[-1])

# Mehrere Werte abrufen
# Syntax: [Start:Ende] -> ACHTUNG: Das ENDE wird nicht mitgenommen!
print(f"Meine ersten zwei Kanäle erhalte ich, indem ich die Indizes 0 und 1 aufrufe: {liste_beispiel[0:2]}")

# Mehrere, nicht zusammen-hängende Werte abrufen
# print(f"Das funktioniert in Python leider NICHT: {liste_beispiel[0,2]}")

# Wir rufen Index 0 und Index 2 einzeln ab und stecken sie händisch in eine neue Liste []
auswahl_basis = [liste_beispiel[0], liste_beispiel[2]]
print(f"Basis-Python Auswahl: {auswahl_basis}")  # Gibt ['Fz', 'Pz']

# Ein flexiblerer Umgang ist möglich, wenn wir mit NumPy Arrays arbeiten. Das folgt in einem späteren Eintrag.
```

```python
# 3. DICTIONARIES (Ungeordnet, über Keys ansprechbar)
# ---------------------------------------------------
# Wenn man wissen will, welche Keys überhaupt existieren:
print(f"Alle Keys im Dictionary: {list(dict_beispiel.keys())}")

# Da es keine Positionen wie 0 oder 1 gibt, rufen wir den Namen des Etiketts auf
print(f"Die Sampling Rate beträgt {dict_beispiel['sfreq']} Hz")
```

In Python laden wir **Bibliotheken (Module)** wie Werkzeugkästen. Mit import X holen wir den ganzen Kasten und müssen danach immer X.werkzeug schreiben. Mit from X import Y greifen wir gezielt in den Kasten X, holen nur das Werkzeug Y (z. B. eine spezifische Klasse oder Funktion) heraus und legen es griffbereit hin, sodass wir direkt Y schreiben können.

```python
# ==========================================
# Bibliotheken importieren: import vs. from
# ==========================================
import mne
from mne_bids import BIDSPath, read_raw_bids

# 1. Was ist ein normaler Import? (Der ganze Werkzeugkasten)
print(f"{type(mne)} - Ein Modul. Das ist der große Werkzeugkasten, der alles enthält.")

# 2. Was haben wir mit 'from ... import ...' herausgeholt?
print(f"{type(BIDSPath)} - Eine Klasse! Wir haben gezielt den Bauplan für BIDS-Pfade auf den Tisch gelegt.")

print(f"{type(read_raw_bids)} - Eine Funktion! (Genau wie in R: Sie nimmt Argumente an und führt eine Aufgabe aus).")
```
